```python
a= [1,2,3,4 ]
labal = ["apple", "mango", "aam", "emli"]
plt.pie(a, labels=labal)
plt.show()
```

```python
a= [1,2,3,4 ]
labal = ["apple", "mango", "aam", "emli"]
plt.pie(a, labels=labal, explode=[0,0.2,0,0])
plt.show()
```

```python
a= [1,2,3,4 ]
labal = ["apple", "mango", "aam", "emli"]
plt.pie(a, labels=labal, explode=[0,0.2,0,0],shadow=True)
plt.show()
```

```python
a= [1,2,3,4 ]
labal = ["apple", "mango", "aam", "emli"]
# c = ['']
colour = ['red','green','blue','yellow']
plt.pie(a, labels=labal, explode=[0,0.2,0,0],shadow=True,autopct = '%1.1f%%',colors=colour)
plt.show()
```

```python
a= [1,2,3,4 ]
labal = ["apple", "mango", "aam", "emli"]
# c = ['']
colour = ['red','green','blue','yellow']
plt.pie(a, labels=labal, explode=[0,0.2,0,0],shadow=True,autopct = '%1.1f%%',colors=colour)
plt.legend()
plt.show()
```
